```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../02_Working_Files/cleaned_call_center_data.csv')
df.head()
```

```python
df.describe()
```

```python
df['performance_tier'].value_counts()
```

```python
df['performance_tier'].value_counts().plot(kind='bar', color='steelblue')
plt.title('Number of Agents by Performance Tier')
plt.xlabel('Performance Tier')
plt.ylabel('Number of Calls')
plt.xticks(rotation=0)
plt.show()

print(df['performance_tier'].value_counts())
```

```python
plt.scatter(df['tenure_months'], df['csat_score'], alpha=0.3, color='seagreen')
plt.title('Tenure vs CSAT Score')
plt.xlabel('Tenure (Months)')
plt.ylabel('CSAT Score')
plt.show()
```

```python
df['tenure_months'].corr(df['csat_score'])
```

```python
df['call_type'].value_counts()
```

```python
plt.scatter(df['qa_audit_score'], df['csat_score'], alpha=0.3, color='darkorange')
plt.title('QA Audit Score vs CSAT Score')
plt.xlabel('QA Audit Score')
plt.ylabel('CSAT Score')
plt.show()
```

```python
df['qa_audit_score'].corr(df['csat_score'])
```

```python
df.groupby('resolution_status')['csat_score'].mean()
```

```python
df.groupby('resolution_status')['csat_score'].mean().plot(kind='bar',color='mediumpurple')
plt.title('Average CSAT Score by Resolution Status')
plt.xlabel('Resolution Status')
plt.ylabel('Average CSAT Score')
plt.xticks(rotation=0)
plt.show()
print(df.groupby('resolution_status')['csat_score'].mean())
```

```python
df.groupby('performance_tier')['aht_minutes'].mean().sort_values()
```

```python
df.groupby('performance_tier')['aht_minutes'].mean().sort_values().plot(kind='bar', color='coral')
plt.title('Average Handle Time by Performance Tier')
plt.xlabel('Performance Tier')
plt.ylabel('Average AHT (Minutes)')
plt.xticks(rotation=0)
plt.show()
print(df.groupby('performance_tier')['aht_minutes'].mean().sort_values())
```
